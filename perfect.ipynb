```python
import os

root_path = "/content/poop_dataset/poop_dataset"

for class_folder in ["Normal", "Loose", "Constipated"]:
    class_path = os.path.join(root_path, class_folder)
    print(f"📂 類別：{class_folder}，圖片數量：{len(os.listdir(class_path))}")
    for f in os.listdir(class_path)[:3]:  # 每類顯示前3張
        print(" →", os.path.join(class_path, f))
```

```python
import os
from ultralytics import YOLO

model = YOLO("weights.pt")

base_input = "/content/poop_dataset/poop_dataset"
output_root = "runs/detect"  # YOLOv8 預設儲存位置

for cls in ["Normal", "Loose", "Constipated"]:
    input_dir = os.path.join(base_input, cls)
    print(f"🔍 預測類別：{cls}...")
    model.predict(source=input_dir, save=True, conf=0.25, name=f"predict_{cls}")
```

```python
from IPython.display import Image, display

# 預覽每類結果中的前 2 張
for cls in ["Normal", "Loose", "Constipated"]:
    print(f"\n📂 類別：{cls}")
    output_dir = f"runs/segment/predict_{cls}"
    files = os.listdir(output_dir)
    for img_file in files[:2]:
        display(Image(filename=os.path.join(output_dir, img_file)))
```
